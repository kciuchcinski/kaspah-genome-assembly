```python
from pathlib import Path

import pandas as pd

TABLES = Path("../pipeline_tables")   # the stage summary tables this repo ships
```

```python
# FOR DRAFT V3 - Hybrid vs illumina sequence gain/loss
# compute per-isolate hybrid_total_len − illumina_total_len and report mean, SD and range

ill_stats = pd.read_csv(TABLES / "assembly_stats.tsv", sep="\t")
ill_stats['library'] = ill_stats['file'].str.split("/").str[1]
ill_stats['strategy'] = ill_stats['file'].str.split("/").str[2]

# Get only illumina asm
ill_stats = ill_stats[ill_stats['strategy'] == 'illumina_unicycler']
```

```python
ill_stats
```

```python
hybrid_stats = pd.read_csv("hybrid_stats.tsv", sep="\t")
```

```python
# remove _polca.fasta suffix (and _v2_polca.fasta) to match illumina library names
hybrid_stats['library'] = hybrid_stats['file'].str.split("/").str[-1].str.replace("_polca.fasta", "").str.replace("_v2_polca.fasta", "")
```

```python
# compute per-isolate hybrid_total_len − illumina_total_len and report mean, SD and range.
len_diff = []

for lib in hybrid_stats['library']:
    hybrid_len = hybrid_stats[hybrid_stats['library'] == lib]['sum_len'].values[0]
    ill_len = ill_stats[ill_stats['library'] == lib]['sum_len'].values[0]
    len_diff.append(hybrid_len - ill_len)

# compute mean, SD and range
import numpy as np

mean_diff = np.mean(len_diff)
std_diff = np.std(len_diff)
range_diff = (np.min(len_diff), np.max(len_diff))

print(f"Mean difference: {mean_diff}")
print(f"Standard deviation: {std_diff}")
print(f"Range: {range_diff}")
```
