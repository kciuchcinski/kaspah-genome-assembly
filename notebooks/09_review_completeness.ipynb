```python
from pathlib import Path

import pandas as pd

TABLES = Path("../pipeline_tables")   # the stage summary tables this repo ships
```

```python
df = pd.read_csv(TABLES / "checkm2_quality_report.tsv", sep="\t")
```

```python
df
```

```python
# get mean, sd, median completeness and contamination
mean_completeness = df["Completeness"].mean()
sd_completeness = df["Completeness"].std()
median_completeness = df["Completeness"].median()
mean_contamination = df["Contamination"].mean()
sd_contamination = df["Contamination"].std()
median_contamination = df["Contamination"].median()

print(f"Mean completeness: {mean_completeness:.2f} ± {sd_completeness:.2f}")
print(f"Median completeness: {median_completeness:.2f}")
print(f"Mean contamination: {mean_contamination:.2f} ± {sd_contamination:.2f}")
print(f"Median contamination: {median_contamination:.2f}")
```
